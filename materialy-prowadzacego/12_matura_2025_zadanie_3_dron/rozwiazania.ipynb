## Rozwiązanie 1: Odczyt i warunki danych

Kod wzorcowy i aktywne testy kontrolne.


````python
def wczytaj(nazwa):
    ruchy = []
    with open(nazwa, "r", encoding="utf-8") as plik:
        for wiersz in plik:
            pola = wiersz.split()
            assert len(pola) == 2
            dx, dy = [int(pole) for pole in pola]
            assert dx > 0
            ruchy.append((dx, dy))
    return ruchy
ruchy_p = wczytaj("dron_przyklad.txt")

# Sprawdzenie
assert len(ruchy_p) == 10
assert ruchy_p[0] == (2000,1001)
assert ruchy_p[-1] == (2000,-1006)
print("Testy zadania 1: OK")
````

## Rozwiązanie 2: 3.1: NWD i zliczanie

Kod wzorcowy i aktywne testy kontrolne.


````python
def nwd(a,b):
    a,b = abs(a),abs(b)
    while b:
        a,b = b,a%b
    return a

def z31(ruchy):
    return sum(nwd(dx,dy) > 1 for dx,dy in ruchy)

# Sprawdzenie
assert z31(ruchy_p) == 6
assert nwd(8,0) == 8
assert nwd(12,-18) == 6
print("Testy zadania 2: OK")
````

## Rozwiązanie 3: Punkty po kolejnych ruchach

Kod wzorcowy i aktywne testy kontrolne.


````python
def trasa(ruchy):
    x = y = 0
    wynik = []
    for dx,dy in ruchy:
        x += dx
        y += dy
        wynik.append((x,y))
    return wynik

# Sprawdzenie
punkty_p = trasa(ruchy_p)
assert punkty_p[:2] == [(2000,1001),(4000,2005)]
assert len(punkty_p) == 10
assert punkty_p[-1] == (20000,0)
print("Testy zadania 3: OK")
````

## Rozwiązanie 4: 3.2 a: wnętrze kwadratu

Kod wzorcowy i aktywne testy kontrolne.


````python
def z32a(punkty):
    return sum(0 < x < 5000 and 0 < y < 5000 for x,y in punkty)

# Sprawdzenie
assert z32a(punkty_p) == 2
assert z32a([(0,1),(5000,1),(1,0),(1,5000)]) == 0
print("Testy zadania 4: OK")
````

## Rozwiązanie 5: 3.2 b: wszystkie kandydatury

Kod wzorcowy i aktywne testy kontrolne.


````python
def z32b(punkty):
    dostepne = set(punkty)
    wynik = []
    for i,a in enumerate(punkty):
        for c in punkty[i+1:]:
            sx,sy = a[0]+c[0],a[1]+c[1]
            if sx%2 == 0 and sy%2 == 0:
                m = (sx//2,sy//2)
                if m in dostepne and m != a and m != c:
                    wynik.append((a,m,c))
    return wynik

# Sprawdzenie
assert z32b(punkty_p) == [((14000,3014),(16000,2010),(18000,1006))]
assert z32b([(0,0),(1,1),(3,3)]) == []
print("Testy zadania 5: OK")
````

## Rozwiązanie 6: Całe zadanie na przykładzie

Kod wzorcowy i aktywne testy kontrolne.


````python
def rozwiaz(ruchy):
    punkty = trasa(ruchy)
    assert punkty[-1] == (20000,0)
    trojki = z32b(punkty)
    assert len(trojki) == 1
    return z31(ruchy),z32a(punkty),trojki[0]

# Sprawdzenie
assert rozwiaz(ruchy_p) == (6,2,((14000,3014),(16000,2010),(18000,1006)))
print("Testy zadania 6: OK")
````

## Rozwiązanie 9: kontrola sześcienna

Kod wzorcowy i aktywne testy kontrolne.


````python
def trojki_wolno(punkty):
    wynik = []
    for i in range(len(punkty)):
        for j in range(i+1,len(punkty)):
            for k in range(j+1,len(punkty)):
                a,m,c = punkty[i],punkty[j],punkty[k]
                if 2*m[0] == a[0]+c[0] and 2*m[1] == a[1]+c[1]:
                    wynik.append((a,m,c))
    return wynik

# Sprawdzenie
assert trojki_wolno(punkty_p) == z32b(punkty_p)
assert trojki_wolno(trasa(ruchy)) == z32b(trasa(ruchy))
print("Testy zadania 9: OK")
````

## Rozwiązanie 10: poprawność fizyczna trasy

Kod wzorcowy i aktywne testy kontrolne.


````python
def poprawny_lot(ruchy):
    if not ruchy or any(dx <= 0 for dx,dy in ruchy):
        return False
    punkty = trasa(ruchy)
    return punkty[-1] == (20000,0) and all(y > 0 for x,y in punkty[:-1])

# Sprawdzenie
assert poprawny_lot(ruchy_p)
assert poprawny_lot(ruchy)
assert not poprawny_lot([(10000,0),(10000,0)])
print("Testy zadania 10: OK")
````
