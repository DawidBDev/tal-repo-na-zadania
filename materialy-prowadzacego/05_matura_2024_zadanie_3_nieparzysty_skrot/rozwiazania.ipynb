````python
n = 39101
wynik = 0
pozycja = 1
while n > 0:
    cyfra = n % 10
    n //= 10
    if cyfra % 2 != 0:
        wynik += cyfra * pozycja
        pozycja *= 10
print(wynik)
assert wynik == 3911
````

## Rozwiązanie 1: 3.1: funkcja zgodna z ograniczeniami

Kod wzorcowy i aktywne testy kontrolne.


````python
def skrot(n):
    wynik = 0
    pozycja = 1
    while n > 0:
        c = n % 10
        n //= 10
        if c % 2 != 0:
            wynik += c * pozycja
            pozycja *= 10
    return wynik

# Sprawdzenie
assert skrot(294762) == 97
assert skrot(39101) == 3911
assert skrot(224) == 0
print("Testy zadania 1: OK")
````

## Rozwiązanie 2: 3.1: przypadki brzegowe

````python
slownik_testow = {n: skrot(n) for n in [1, 2, 9, 10101, 20003, 86420]}

# Sprawdzenie
assert slownik_testow == {1: 1, 2: 0, 9: 9, 10101: 111, 20003: 3, 86420: 0}
print("Testy zadania 2: OK")
````

## Rozwiązanie 3: Wczytanie obu przykładów

Kod wzorcowy i aktywne testy kontrolne.


````python
def wczytaj(nazwa):
    liczby = []
    with open(nazwa, "r", encoding="utf-8") as plik:
        for wiersz in plik:
            liczby.append(int(wiersz))
    return liczby
przyklad_1 = wczytaj("skrot_przyklad.txt")
przyklad_2 = wczytaj("skrot2_przyklad.txt")

# Sprawdzenie
assert len(przyklad_1) == len(przyklad_2) == 20
assert all(0 < n < 30000 for n in przyklad_1 + przyklad_2)
print("Testy zadania 3: OK")
````

## Rozwiązanie 4: 3.2: liczność i największy element

Kod wzorcowy i aktywne testy kontrolne.


````python
def bez_skrotu(liczby):
    ile = 0
    najwieksza = None
    for n in liczby:
        if skrot(n) == 0:
            ile += 1
            if najwieksza is None or n > najwieksza:
                najwieksza = n
    return ile, najwieksza

# Sprawdzenie
assert bez_skrotu(przyklad_1) == (2, 2428)
assert bez_skrotu([135, 97]) == (0, None)
print("Testy zadania 4: OK")
````

## Rozwiązanie 5: 3.3: funkcja NWD

Kod wzorcowy i aktywne testy kontrolne.


````python
def nwd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Sprawdzenie
assert skrot(4872) == 7
assert nwd(4872, skrot(4872)) == 7
assert nwd(21, 63) == 21
print("Testy zadania 5: OK")
````

## Rozwiązanie 6: 3.3: lista odpowiedzi

Kod wzorcowy i aktywne testy kontrolne.


````python
def nwd_siedem(liczby):
    wynik = []
    for n in liczby:
        m = skrot(n)
        if m != 0 and nwd(n, m) == 7:
            wynik.append(n)
    return wynik

# Sprawdzenie
assert nwd_siedem(przyklad_2) == [4872, 23527]
assert nwd_siedem([4872, 4872]) == [4872, 4872]
print("Testy zadania 6: OK")
````

## Rozwiązanie 7: Obliczenia na pełnych danych

## Rozwiązanie 9: niezależny wzorzec testowy

Kod wzorcowy i aktywne testy kontrolne.


````python
def skrot_testowy(n):
    zapis = "".join(c for c in str(n) if int(c) % 2 == 1)
    return int(zapis) if zapis else 0

# Sprawdzenie
for n in range(1, 1000):
    assert skrot(n) == skrot_testowy(n)
print("Testy zadania 9: OK")
````

## Rozwiązanie 10: własność skrótu

Kod wzorcowy i aktywne testy kontrolne.


````python
idempotentny = all(skrot(skrot(n)) == skrot(n) for n in range(1, 1000) if skrot(n) != 0)

# Sprawdzenie
assert idempotentny
print("Testy zadania 10: OK")
````
